```python
from vllm import LLM, SamplingParams
import io
import os
import sys

selected_option = dropdown.value

class FilenoFix:
    """Wraps a stream and exposes a real OS-level fileno(),
    working around ipykernel's OutStream not supporting it."""
    def __init__(self, stream, fd):
        self._stream = stream
        self._fd = fd
    def __getattr__(self, name):
        return getattr(self._stream, name)
    def fileno(self):
        return self._fd
    def writable(self):
        return True

"""Applies the fileno() patch only if it is actually broken."""
try:
    sys.stdout.fileno()
except (io.UnsupportedOperation, AttributeError):
    sys.stdout = FilenoFix(sys.stdout, 1)

try:
    sys.stderr.fileno()
except (io.UnsupportedOperation, AttributeError):
    sys.stderr = FilenoFix(sys.stderr, 2)

# 3. Load and return the model
print(f"Loading {selected_option} into GPU memory...")

llm = LLM(model=selected_option,
          max_model_len=2048,
          enforce_eager=True)
```

```python
"""
Replaces free-text parsing with vLLM's guided decoding: the model is
constrained to output ONLY one of the item's valid response codes, so
there is no more "which number in this paragraph was the real answer"
ambiguity. This directly fixes the Q164/EGY issue shown in the raw text.

Note: guided decoding syntax has changed across vLLM versions. This uses
the vLLM v0.12+ API (StructuredOutputsParams). If you're on an older vLLM
version and this import fails, tell me your `vllm.__version__` — older
versions use `GuidedDecodingParams` with `guided_decoding=` instead.
"""
import csv
from vllm import LLM, SamplingParams
from vllm.sampling_params import StructuredOutputsParams
#from item_bank import ITEMS

#PILOT_COUNTRIES = ["EGY", "NLD", "CHN", "KOR", "USA"]
#COUNTRY_NAMES = {"EGY": "Egypt", "NLD": "the Netherlands", "CHN": "China",
#                  "KOR": "South Korea", "USA": "the United States"}

PILOT_COUNTRIES = ["EGY", "NLD", "CHN", "KOR", "USA", "GBR", "DEU",
                  "RUS", "UKR", "MEX", "ARG", "JPN", "NGA", "IDN", "IND"]
COUNTRY_NAMES = {
    "USA": "the United States",
    "GBR": "Great Britain",
    "DEU": "Germany",
    "NLD": "the Netherlands",
    "RUS": "Russia",
    "UKR": "Ukraine",
    "MEX": "Mexico",
    "ARG": "Argentina",
    "CHN": "China",
    "JPN": "Japan",
    "KOR": "South Korea",
    "EGY": "Egypt",
    "NGA": "Nigeria",
    "IDN": "Indonesia",
    "IND": "India"
}

N_SAMPLES = 20
TEMPERATURE = 0.8
OUT_PATH = f"/content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions/{selected_option.split("/")[0]}/llm_responses.csv"

#llm = LLM(model=MODEL)


def format_options(item):
    if "anchors" in item:
        lo_code, lo_label, hi_code, hi_label = item["anchors"]
        return f"Scale: {lo_code} ({lo_label}) to {hi_code} ({hi_label})."
    return "\n".join(f"{code}. {label}" for code, label in item["options"])


def build_prompt(item, country: str, condition: str) -> str:
    name = COUNTRY_NAMES[country]
    persona = (
        f"Answer as a typical resident of {name} would answer this survey question. "
        f"Respond the way you believe most people in {name} would, not your own personal view.\n\n"
        if condition == "cultural" else ""
    )
    return f"{persona}{item['question']}\n\n{format_options(item)}\n\nRespond with only the number."

tasks, prompts, sampling_params_list = [], [], []

# 1. Default (unprompted) baseline - generated once per item across N_SAMPLES
for country in PILOT_COUNTRIES:
  for item in ITEMS:
    valid_choices = [str(c) for c, _ in item["options"]]
    tasks.append((country, item, "default"))
    prompts.append(build_prompt(item, country, condition="default"))
    sampling_params_list.append(
        SamplingParams(
            temperature=TEMPERATURE,
            max_tokens=5,
            n=N_SAMPLES,
            structured_outputs=StructuredOutputsParams(choice=valid_choices),
        )
    )

# 2. Culturally steered conditions - generated per country
for country in PILOT_COUNTRIES:
  for item in ITEMS:
    valid_choices = [str(c) for c, _ in item["options"]]
    tasks.append((country, item, "cultural"))
    prompts.append(build_prompt(item, country, condition="cultural"))
    sampling_params_list.append(
      SamplingParams(
          temperature=TEMPERATURE,
          max_tokens=5,
          n=N_SAMPLES,
          structured_outputs=StructuredOutputsParams(choice=valid_choices),
      )
  )

print(f"Sending {len(prompts)} prompts x {N_SAMPLES} samples each to vLLM (guided decoding)...")
# .chat() applies Qwen's chat template (system/user turns) instead of raw strings --
# matters more for a 1.5B model than a larger one, since it has less capacity to
# generalize past a prompt shape it wasn't instruction-tuned on.
conversations = [[{"role": "user", "content": p}] for p in prompts]
outputs = llm.chat(conversations, sampling_params_list)

rows = []
for (country, item, condition), output in zip(tasks, outputs):
    for completion in output.outputs:
        raw = completion.text.strip()
        parsed = int(raw) if raw.lstrip("-").isdigit() else None  # should always succeed under guided decoding
        rows.append({
            "country": country, "col": item["col"], "dim": item["dim"],
            "sub": item["sub"], "condition": condition,
            "raw_response": raw, "parsed_code": parsed,
        })

with open(OUT_PATH, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=rows[0].keys())
    writer.writeheader()
    writer.writerows(rows)

n_unparsed = sum(1 for r in rows if r["parsed_code"] is None)
print(f"Saved {len(rows)} responses to {OUT_PATH}")
print(f"Unparsed responses: {n_unparsed} ({n_unparsed/len(rows):.1%}) — should be ~0% now")






# /content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions (1)
```

```python
"""
Step 5: reconstruct SACSECVAL / RESEMAVAL from LLM parsed_code responses,
per country x condition, then compare against the human WVS baseline.

Simplification vs. the official Welzel formula: each item is linearly
rescaled 0-1 in its emancipative/secular direction and averaged within its
sub-index; sub-indices are then averaged into the dimension score. The
official formula uses population-level quintile rescaling we can't
replicate from a 5-country pilot -- state this as a scope decision.
"""
import pandas as pd
import numpy as np
from scipy import stats

LLM_PATH = f"/content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions/{selected_option.split("/")[0]}/llm_responses.csv"                          # your elicitation output
HUMAN_PATH = "/content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions/pilot_countries.csv"  # human baseline, pilot subset

# --- item-level rescaling rules: col -> function(code) -> 0..1 in the emancipative/secular direction
def linear(lo, hi, invert=False):
    def f(v):
        if pd.isna(v):
            return np.nan
        x = (v - lo) / (hi - lo)
        return 1 - x if invert else x
    return f

ITEM_SCORING = {
    # RESEMAVAL: AUTONOMY (binary: 1=Important, 2=Not mentioned)
    "Q8P":  {"sub": "AUTONOMY", "fn": lambda v: {1: 1.0, 2: 0.0}.get(v, np.nan)},
    "Q11P": {"sub": "AUTONOMY", "fn": lambda v: {1: 1.0, 2: 0.0}.get(v, np.nan)},
    "Q17P": {"sub": "AUTONOMY", "fn": lambda v: {1: 0.0, 2: 1.0}.get(v, np.nan)},  # obedience: inverted

    # RESEMAVAL: EQUALITY
    "Q33P": {"sub": "EQUALITY", "fn": linear(1, 5)},   # 1=agree(men priority,low eq) .. 5=disagree strongly(high eq)
    "Q29P": {"sub": "EQUALITY", "fn": linear(1, 4)},
    "Q30P": {"sub": "EQUALITY", "fn": linear(1, 4)},

    # RESEMAVAL: CHOICE (1=never justifiable .. 10=always justifiable)
    "Q182": {"sub": "CHOICE", "fn": linear(1, 10)},
    "Q184": {"sub": "CHOICE", "fn": linear(1, 10)},
    "Q185": {"sub": "CHOICE", "fn": linear(1, 10)},

    # RESEMAVAL: VOICE (forced choice -> 1 if postmaterialist option chosen, else 0)
    "Q152": {"sub": "VOICE", "fn": lambda v: {1: 0.0, 2: 0.0, 3: 1.0, 4: 0.0}.get(v, np.nan)},
    "Q154": {"sub": "VOICE", "fn": lambda v: {1: 0.0, 2: 1.0, 3: 0.0, 4: 1.0}.get(v, np.nan)},

    # SACSECVAL: DEFIANCE
    "Q45P":  {"sub": "DEFIANCE", "fn": linear(1, 3)},   # 1=good thing(low defiance) .. 3=bad thing(high)
    "Q254P": {"sub": "DEFIANCE", "fn": lambda v: linear(1, 4)(v) if v in (1, 2, 3, 4) else np.nan},
    "Q164":  {"sub": "DEFIANCE", "fn": linear(1, 10, invert=True)},  # importance of God, inverted

    # SACSECVAL: DISBELIEF
    "Q6P":   {"sub": "DISBELIEF", "fn": linear(1, 4)},
    "Q173P": {"sub": "DISBELIEF", "fn": linear(1, 3)},
    "Q171P": {"sub": "DISBELIEF", "fn": linear(1, 7)},

    # SACSECVAL: RELATIVISM
    "Q178": {"sub": "RELATIVISM", "fn": linear(1, 10)},
    "Q180": {"sub": "RELATIVISM", "fn": linear(1, 10)},
    "Q181": {"sub": "RELATIVISM", "fn": linear(1, 10)},

    # SACSECVAL: SCEPTICISM
    "Q65P": {"sub": "SCEPTICISM", "fn": linear(1, 4)},
    "Q69P": {"sub": "SCEPTICISM", "fn": linear(1, 4)},
    "Q70P": {"sub": "SCEPTICISM", "fn": linear(1, 4)},
}

DIM_OF_SUB = {
    "AUTONOMY": "RESEMAVAL", "EQUALITY": "RESEMAVAL", "CHOICE": "RESEMAVAL", "VOICE": "RESEMAVAL",
    "DEFIANCE": "SACSECVAL", "DISBELIEF": "SACSECVAL", "RELATIVISM": "SACSECVAL", "SCEPTICISM": "SACSECVAL",
}


def score_llm(df: pd.DataFrame) -> pd.DataFrame:
    df = df.dropna(subset=["parsed_code"]).copy()
    df["item_score"] = df.apply(lambda r: ITEM_SCORING[r["col"]]["fn"](r["parsed_code"]), axis=1)
    df["sub"] = df["col"].map(lambda c: ITEM_SCORING[c]["sub"])
    df["target_dim"] = df["sub"].map(DIM_OF_SUB)

    # mean per respondent-draw is meaningless here (each row is one sampled draw);
    # aggregate straight to sub-index mean per country x condition
    sub_scores = (df.groupby(["country", "condition", "target_dim", "sub"])["item_score"]
                  .mean().reset_index())
    dim_scores = (sub_scores.groupby(["country", "condition", "target_dim"])["item_score"]
                  .mean().reset_index()
                  .pivot(index=["country", "condition"], columns="target_dim", values="item_score")
                  .reset_index())
    return dim_scores


def main():
    llm_raw = pd.read_csv(LLM_PATH)
    llm_scores = score_llm(llm_raw)
    human = pd.read_csv(HUMAN_PATH)[["country", "SACSECVAL", "RESEMAVAL"]]
    human = human.rename(columns={"SACSECVAL": "human_SACSECVAL", "RESEMAVAL": "human_RESEMAVAL"})

    merged = llm_scores.merge(human, on="country")
    merged.to_csv(f"/content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions/{selected_option.split("/")[0]}/{selected_option.split("/")[0]}_alignment_scores.csv", index=False)
    print(merged.to_string(index=False))

    print("\n--- Correlation: LLM vs human, by condition ---")
    for cond in ["default", "cultural"]:
        sub = merged[merged["condition"] == cond]
        for dim in ["SACSECVAL", "RESEMAVAL"]:
            r, p = stats.spearmanr(sub[dim], sub[f"human_{dim}"])
            mae = (sub[dim] - sub[f"human_{dim}"]).abs().mean()
            print(f"{cond:9s} {dim:10s} Spearman r={r:.3f} (p={p:.3f})  MAE={mae:.3f}")

    print("\n--- Does cultural prompting reduce error vs default? (paired, per country) ---")
    for dim in ["SACSECVAL", "RESEMAVAL"]:
        piv = merged.pivot(index="country", columns="condition", values=dim)
        human_col = merged.drop_duplicates("country").set_index("country")[f"human_{dim}"]
        err_default = (piv["default"] - human_col).abs()
        err_cultural = (piv["cultural"] - human_col).abs()
        stat, p = stats.wilcoxon(err_default, err_cultural)
        print(f"{dim}: mean|err| default={err_default.mean():.3f} cultural={err_cultural.mean():.3f} "
              f"Wilcoxon p={p:.3f}")


if __name__ == "__main__":
    main()
```

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. Load experiment scores
df = pd.read_csv(f"/content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions/{selected_option.split("/")[0]}/alignment_scores.csv")

# 2. Pivot model scores and extract unique human benchmarks
piv_res = df.pivot(index="country", columns="condition", values="RESEMAVAL")
piv_sac = df.pivot(index="country", columns="condition", values="SACSECVAL")
human = df.drop_duplicates("country").set_index("country")

# 3. Calculate 2D Euclidean distances
dist_def = np.sqrt(
    (piv_res["default"] - human["human_RESEMAVAL"]) ** 2
    + (piv_sac["default"] - human["human_SACSECVAL"]) ** 2
)

dist_cul = np.sqrt(
    (piv_res["cultural"] - human["human_RESEMAVAL"]) ** 2
    + (piv_sac["cultural"] - human["human_SACSECVAL"]) ** 2
)

# 4. Compute delta and outcome
delta_dist = dist_def - dist_cul
outcome = np.where(delta_dist > 0, "Improved", "Worsened")

# 5. Build summary DataFrame
summary_df = pd.DataFrame(
    {
        "Country": dist_def.index,
        "d_default": dist_def.values,
        "d_cultural": dist_cul.values,
        "Reduction (d_def - d_cul)": delta_dist.values,
        "Outcome in Joint Space": outcome,
    }
)

# Sort by country or reduction as desired
summary_df = summary_df.sort_values(
    by="Reduction (d_def - d_cul)", ascending=False
).reset_index(drop=True)

# 6. Compute summary statistics
mean_row = pd.DataFrame(
    [
        {
            "Country": "Mean",
            "d_default": dist_def.mean(),
            "d_cultural": dist_cul.mean(),
            "Reduction (d_def - d_cul)": delta_dist.mean(),
            "Outcome in Joint Space": (
                f"{sum(delta_dist > 0)} Improved, {sum(delta_dist < 0)} Worsened"
            ),
        }
    ]
)

final_table = pd.concat([summary_df, mean_row], ignore_index=True)

# Format floats to 4 decimal places
formatted_table = final_table.copy()
for col in ["d_default", "d_cultural", "Reduction (d_def - d_cul)"]:
    formatted_table[col] = formatted_table[col].apply(
        lambda x: f"{x:+.4f}" if col.startswith("Reduction") else f"{x:.4f}"
    )

print(formatted_table.to_string(index=False))

# 7. Statistical significance
stat, p_val = stats.wilcoxon(dist_def, dist_cul)
print(f"\nPaired Wilcoxon Signed-Rank Test: stat={stat}, p-value={p_val:.4f}")

# Export to CSV
formatted_table.to_csv(f"/content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions/{selected_option.split("/")[0]}/{selected_option.split("/")[0]}_euclidean_distance_summary.csv", index=False)
```
