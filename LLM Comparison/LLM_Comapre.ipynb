```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Empirical benchmark outcomes from cross-model evaluation
models = [
    "Qwen",
    "Meta Llama",
    "Google Gemma",
    "Mistral",
    "Hugging Face",
]

# Spearman correlations under cultural condition
r_resem = [0.836, 0.425, 0.707, 0.614, 0.400]
r_sacsec = [0.250, 0.482, 0.672, 0.554, -0.164]

# Net 2D Euclidean Distance Reduction (Delta d = d_def - d_cul)
delta_2d = [
    0.3699 - 0.2970,  # +0.0729
    0.3017 - 0.1403,  # +0.1614
    0.3190 - 0.1271,  # +0.1919
    0.4039 - 0.1437,  # +0.2602
    0.1519 - 0.1578,  # -0.0059
]

# Dispersion Ratios under cultural condition (sigma_LLM / sigma_Human)
disp_resem = [0.52, 0.59, 0.70, 0.44, 0.28]
disp_sacsec = [0.34, 0.43, 0.58, 0.50, 0.30]

x = np.arange(len(models))
width = 0.35

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5.5))

# --- Panel 1: Rank Alignment (Spearman r) ---
b1 = ax1.bar(
    x - width / 2,
    r_resem,
    width,
    label="Emancipative (RESEM)",
    color="#2166AC",
    alpha=0.9,
)
b2 = ax1.bar(
    x + width / 2,
    r_sacsec,
    width,
    label="Secular (SACSEC)",
    color="#B2182B",
    alpha=0.85,
)
ax1.axhline(0, color="black", linestyle="-", linewidth=0.8)
ax1.set_ylabel("Spearman Rank Correlation ($r_s$)", fontsize=11)
ax1.set_title(
    "A. Dimensional Rank Alignment (Cultural)", fontsize=12, fontweight="bold"
)
ax1.set_xticks(x)
ax1.set_xticklabels(models, fontsize=9.5)
ax1.set_ylim(-0.3, 1.0)
ax1.legend(loc="upper right", frameon=True)
ax1.grid(axis="y", linestyle=":", alpha=0.6)

# --- Panel 2: Net 2D Geometric Shift ---
colors_2d = ["#2A9D8F" if d > 0 else "#E76F51" for d in delta_2d]
bars_2d = ax2.bar(
    x, delta_2d, width=0.55, color=colors_2d, edgecolor="black", linewidth=0.5
)
ax2.axhline(0, color="black", linestyle="--", linewidth=0.8)
ax2.set_ylabel(
    "Net 2D Distance Reduction ($\Delta d = \overline{d}_{def} - \overline{d}_{cul}$)",
    fontsize=11,
)
ax2.set_title(
    "B. Holistic 2D Spatial Convergence", fontsize=12, fontweight="bold"
)
ax2.set_xticks(x)
ax2.set_xticklabels(models, fontsize=9.5)
ax2.grid(axis="y", linestyle=":", alpha=0.6)

for bar in bars_2d:
  yval = bar.get_height()
  va = "bottom" if yval >= 0 else "top"
  offset = 0.008 if yval >= 0 else -0.015
  ax2.text(
      bar.get_x() + bar.get_width() / 2.0,
      yval + offset,
      f"{yval:+.3f}",
      ha="center",
      va=va,
      fontsize=9,
      fontweight="bold",
  )

# --- Panel 3: Variance Dispersion Ratio ---
b3 = ax3.bar(
    x - width / 2,
    disp_resem,
    width,
    label="Emancipative Ratio",
    color="#4393C3",
    alpha=0.9,
)
b4 = ax3.bar(
    x + width / 2,
    disp_sacsec,
    width,
    label="Secular Ratio",
    color="#D6604D",
    alpha=0.9,
)
ax3.axhline(
    1.0,
    color="darkgreen",
    linestyle="--",
    linewidth=1.0,
    label="Human Benchmark (1.0)",
)
ax3.set_ylabel(
    "Dispersion Ratio ($\sigma_{LLM} / \sigma_{Human}$)", fontsize=11
)
ax3.set_title("C. Variance Preservation vs. Collapse", fontsize=12, fontweight="bold")
ax3.set_xticks(x)
ax3.set_xticklabels(models, fontsize=9.5)
ax3.set_ylim(0, 1.15)
ax3.legend(loc="upper right", frameon=True)
ax3.grid(axis="y", linestyle=":", alpha=0.6)

plt.suptitle(
    "Cross-Architectural Model Comparison (Welzel Cultural Dimensions)",
    fontsize=14,
    y=1.02,
)
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions/LLM Comparison/cross_model_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Architectures and their joint 2D Euclidean distances to human ground truth
data = {
    "Mistral-Instruct-v0.3 (7B)": {
        "def": 0.4039,
        "cul": 0.1437,
        "imp": "15 / 0",
    },
    "Google Gemma 2-it (2B)": {"def": 0.3190, "cul": 0.1271, "imp": "13 / 2"},
    "Meta Llama 3 (3B/8B)": {"def": 0.3017, "cul": 0.1403, "imp": "11 / 4"},
    "Qwen 2.5-Instruct (1.5B)": {
        "def": 0.3699,
        "cul": 0.2970,
        "imp": "15 / 0",
    },
    "SmolLM2-Instruct (1.7B)": {"def": 0.1519, "cul": 0.1578, "imp": "5 / 10"},
}

models = list(data.keys())
y_pos = np.arange(len(models))

d_def = [data[m]["def"] for m in models]
d_cul = [data[m]["cul"] for m in models]

plt.figure(figsize=(9, 5))

# Draw connecting convergence/divergence lines
for i, m in enumerate(models):
  color = "#2A9D8F" if d_cul[i] < d_def[i] else "#E76F51"
  plt.plot([d_def[i], d_cul[i]], [i, i], color=color, linewidth=2.5, zorder=2)
  # Annotate country win/loss record
  plt.text(
      max(d_def[i], d_cul[i]) + 0.015,
      i,
      f"Improved: {data[m]['imp']}",
      va="center",
      fontsize=8.5,
      color="#333333",
  )

plt.scatter(
    d_def,
    y_pos,
    color="#777777",
    s=70,
    label="Default Error ($\overline{d}_{def}$)",
    zorder=3,
)
plt.scatter(
    d_cul,
    y_pos,
    color="#2166AC",
    s=70,
    label="Cultural Error ($\overline{d}_{cul}$)",
    zorder=3,
)

plt.yticks(y_pos, models, fontsize=10)
plt.xlabel("Mean 2D Euclidean Distance to Human Ground Truth", fontsize=11)
plt.title(
    "Cross-Architectural 2D Cultural Calibration (Default vs. Cultural)",
    fontsize=12,
    pad=12,
)
plt.xlim(0.08, 0.48)
plt.grid(axis="x", linestyle=":", alpha=0.6)
plt.legend(loc="upper right", frameon=True)

plt.tight_layout()
plt.savefig("/content/drive/MyDrive/Colab Notebooks/AdvanceTopics/AdvanceTopics - Evaluating the Robustness of Cultural Prompting in Large Language Models Against Human Cultural Value Distributions/LLM Comparison/cross_model_2d_dumbbell.png", dpi=300)
plt.show()
```
